## 1️⃣ Task 1: Food Delivery Orders & Initial Pairwise Distance Matrix

### 📖 Conceptual Definition & Mathematical Foundations
- **Agglomerative Hierarchical Clustering** begins at the bottom of the hierarchy where every single data point is its own singleton cluster ($N$ clusters).
- **The Distance Matrix ($D \in \mathbb{R}^{N \times N}$):** A symmetric square matrix containing the pairwise distances between all observations. It serves as the fundamental input to determine which two clusters are closest and should be merged first.
- **Properties of a Distance Matrix:**
  1. **Self-Distance is Zero:** $D_{ii} = 0$ for all $i$.
  2. **Symmetry:** $D_{ij} = D_{ji}$ (distance from $A$ to $B$ equals distance from $B$ to $A$).
  3. **Non-negativity:** $D_{ij} \ge 0$.
  4. **Number of Unique Pairs:** For $N = 6$ orders, there are $\binom{6}{2} = \frac{6 \times 5}{2} = 15$ unique distance values.

## 2️⃣ Task 2: Agglomerative Hierarchical Clustering on 8 Songs & Dendrogram

### 📖 Conceptual Definition: Dendrogram & Tree Hierarchy
- **Hierarchical Tree Structure (Dendrogram):** A tree diagram that records the sequential merges of clusters at each iteration step.
- **Leaves:** Individual song observations located at vertical height $y = 0$.
- **Inverted U-Bars:** Represent the merging of two branches. The vertical height of the horizontal bar indicates the **linkage distance** between the two clusters being combined.
- **Horizontal Cut:** Cutting across the dendrogram at a specific threshold distance determines the final number of discrete clusters $k$.

---

### 🎵 The 8 Songs Dataset (Tempo & Energy Profile):
1. **S1: Acoustic Morning** (Tempo: 70 BPM, Energy: 0.20) — Chill Acoustic
2. **S2: Lofi Midnight** (Tempo: 75 BPM, Energy: 0.25) — Lo-Fi Chillhop
3. **S3: Coffeehouse Guitar** (Tempo: 82 BPM, Energy: 0.32) — Indie Folk
4. **S4: Retro Pop Dance** (Tempo: 118 BPM, Energy: 0.70) — Nu-Disco / Pop
5. **S5: Synthwave Cruiser** (Tempo: 124 BPM, Energy: 0.78) — Synthwave
6. **S6: Festival Anthem** (Tempo: 128 BPM, Energy: 0.88) — Mainstage EDM
7. **S7: Hard Dance Power** (Tempo: 155 BPM, Energy: 0.94) — Hardstyle
8. **S8: Speedcore Rave** (Tempo: 172 BPM, Energy: 0.98) — Industrial Techno

```python
# =====================================================================
# TASK 2: Agglomerative Clustering on 8 Songs & Scipy Dendrogram
# =====================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# 1. Define the 8 songs dataset with Tempo (BPM) and Energy (0 to 1)
songs_data = pd.DataFrame([
    {"song_id": "S1", "title": "Acoustic Morning", "genre": "Acoustic", "tempo": 70, "energy": 0.20},
    {"song_id": "S2", "title": "Lofi Midnight", "genre": "Lo-Fi", "tempo": 75, "energy": 0.25},
    {"song_id": "S3", "title": "Coffeehouse Guitar", "genre": "Indie Folk", "tempo": 82, "energy": 0.32},
    {"song_id": "S4", "title": "Retro Pop Dance", "genre": "Pop", "tempo": 118, "energy": 0.70},
    {"song_id": "S5", "title": "Synthwave Cruiser", "genre": "Synthwave", "tempo": 124, "energy": 0.78},
    {"song_id": "S6", "title": "Festival Anthem", "genre": "EDM", "tempo": 128, "energy": 0.88},
    {"song_id": "S7", "title": "Hard Dance Power", "genre": "Hardstyle", "tempo": 155, "energy": 0.94},
    {"song_id": "S8", "title": "Speedcore Rave", "genre": "Techno", "tempo": 172, "energy": 0.98}
])

# 2. Standardize features to equalize Tempo (70-172) and Energy (0.2-1.0)
features = ["tempo", "energy"]
scaler = StandardScaler()
X_songs_scaled = scaler.fit_transform(songs_data[features])

# 3. Compute hierarchical linkage using Scipy (Ward method)
linkage_matrix = linkage(X_songs_scaled, method='ward')

# 4. Fit Scikit-Learn AgglomerativeClustering for k=3 clusters
agg_cluster = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
songs_data["cluster"] = agg_cluster.fit_predict(X_songs_scaled)

# 5. Plot the Dendrogram
fig, ax = plt.subplots(figsize=(10, 6))
labels_for_dendro = [f"{row['song_id']}: {row['title']} ({row['genre']})" for _, row in songs_data.iterrows()]

dendrogram(
    linkage_matrix,
    labels=labels_for_dendro,
    orientation='top',
    distance_sort='descending',
    show_leaf_counts=True,
    ax=ax,
    color_threshold=2.5  # Colors branches below threshold 2.5
)

# Draw threshold line showing 3 natural clusters cut
ax.axhline(y=2.5, color='#e74c3c', linestyle='--', linewidth=1.8, label='Cluster Cutoff Threshold (k=3)')

ax.set_title("Agglomerative Hierarchical Clustering: Song Dendrogram (Ward Linkage)",
             fontsize=13, fontweight='bold', pad=12)
ax.set_ylabel("Linkage Euclidean Distance (Merge Height)", fontsize=11, fontweight='bold')
ax.set_xlabel("Song Tracks", fontsize=11, fontweight='bold')
plt.xticks(rotation=35, ha='right', fontsize=9.5, fontweight='semibold')
ax.grid(True, linestyle=":", alpha=0.6)
ax.legend(loc="upper right", frameon=True, facecolor="#fdfefe")

plt.tight_layout()
plt.show()
```

## 3️⃣ Task 3: Comparing Linkage Methods (`single`, `complete`, `average`)

### 📖 Conceptual Definition of Linkage Criteria
How do we measure the distance between two multi-point clusters $A$ and $B$?

1. **Single Linkage (Minimum Distance / Nearest Neighbor):**
   $$D_{\text{single}}(A, B) = \min_{x \in A, y \in B} d(x, y)$$
   - Merges clusters based on their closest pair of points.
   - **Characteristics:** Prone to the **chaining effect** (forming elongated, snake-like clusters connected by occasional bridging points). Sensitive to noise.

2. **Complete Linkage (Maximum Distance / Furthest Neighbor):**
   $$D_{\text{complete}}(A, B) = \max_{x \in A, y \in B} d(x, y)$$
   - Merges clusters whose furthest pair of points is smallest.
   - **Characteristics:** Enforces tight, compact, spherical clusters with small diameters. Highly sensitive to outliers.

3. **Average Linkage (UPGMA - Mean Pairwise Distance):**
   $$D_{\text{average}}(A, B) = \frac{1}{|A| |B|} \sum_{x \in A} \sum_{y \in B} d(x, y)$$
   - Computes the average distance between all pairs of points across both clusters.
   - **Characteristics:** Balanced, robust compromise between the extremes of single and complete linkage.

```python
# =====================================================================
# TASK 3: Side-by-Side Comparison of Single, Complete, and Average Linkage
# =====================================================================
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

linkage_methods = ["single", "complete", "average"]
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

short_labels = [f"{r['song_id']}: {r['title'][:10]}" for _, r in songs_data.iterrows()]

for idx, method in enumerate(linkage_methods):
    Z = linkage(X_songs_scaled, method=method)
    ax = axes[idx]
    
    dendrogram(
        Z,
        labels=short_labels,
        orientation='top',
        ax=ax,
        color_threshold=0.7 * max(Z[:, 2])
    )
    
    ax.set_title(f"Method: {method.capitalize()} Linkage", fontsize=12, fontweight='bold', pad=10)
    ax.set_ylabel("Distance" if idx == 0 else "", fontsize=10, fontweight='bold')
    ax.set_xlabel("Songs", fontsize=10, fontweight='bold')
    ax.tick_params(axis='x', rotation=45)
    ax.grid(True, linestyle=":", alpha=0.5)

plt.suptitle("Structural Comparison of Dendrograms Across Different Linkage Criteria",
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

### 🔍 Detailed Observations on Structural Changes:

| Linkage Method | Merge Distance Scale | Chaining Tendency | Resulting Cluster Shapes | Best Real-World Use Case |
| :--- | :---: | :---: | :--- | :--- |
| **Single (Minimum)** | Very small (merges at lowest thresholds) | **High (Severe Chaining):** Successively absorbs individual points one-by-one into a long chain. | Elongated, non-convex, continuous streams. | Tracing physical geographical rivers, transit lines, or non-linear manifolds. |
| **Complete (Maximum)** | High (merges at larger distances) | **Zero (Compactness Bias):** Refuses to merge until all outer points are close. | Tightly bounded, uniform, spherical clusters with small diameters. | Strict customer segmentation, quality control where diameter must be constrained. |
| **Average (Mean)** | Intermediate (smooth balanced heights) | **Low (Balanced Compromise):** Weights all pairwise interactions evenly. | Natural, moderately compact clusters; robust against isolated noisy tracks. | General taxonomy building, gene expression, music audio similarity profiling. |

## 4️⃣ Task 4: Dual Distance Function (Euclidean & Manhattan)

### 📖 Conceptual Definition
Given two points in 2D space $P_1(x_1, y_1)$ and $P_2(x_2, y_2)$:

1. **Euclidean Distance ($L_2$ Norm):** The straight-line hypotenuse distance:
   $$d_{\text{Euclidean}} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
2. **Manhattan Distance ($L_1$ Norm):** The rectilinear grid travel distance:
   $$d_{\text{Manhattan}} = |x_2 - x_1| + |y_2 - y_1|$$

*Geometric Property:* By the triangle inequality, $d_{\text{Euclidean}} \le d_{\text{Manhattan}}$ for any pair of coordinates, with equality holding if and only if points are aligned along a single cardinal axis.

```python
# =====================================================================
# TASK 4: Dual Distance Function & Testing on 3 Coordinate Pairs
# =====================================================================
import math
import pandas as pd

def calculate_distances(p1, p2):
    """
    Computes both Euclidean (L2) and Manhattan (L1) distances between two 2D points.
    
    Parameters:
        p1 (tuple/list): (x1, y1)
        p2 (tuple/list): (x2, y2)
        
    Returns:
        dict: {'euclidean': float, 'manhattan': float}
    """
    x1, y1 = p1
    x2, y2 = p2
    
    euclidean = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    manhattan = abs(x2 - x1) + abs(y2 - y1)
    
    return {
        "euclidean": round(euclidean, 4),
        "manhattan": round(manhattan, 4)
    }


# Test on 3 distinct coordinate pairs from the delivery orders dataset
# O1: (15, 2), O2: (20, 3), O4: (40, 8), O6: (55, 14)
test_pairs = [
    {"label": "Pair 1 (O1 vs O2)", "p1": (15, 2), "p2": (20, 3)},
    {"label": "Pair 2 (O1 vs O4)", "p1": (15, 2), "p2": (40, 8)},
    {"label": "Pair 3 (O3 vs O6)", "p1": (35, 7), "p2": (55, 14)}
]

test_results = []
for test in test_pairs:
    dists = calculate_distances(test["p1"], test["p2"])
    test_results.append({
        "Comparison": test["label"],
        "Point 1 (x1, y1)": str(test["p1"]),
        "Point 2 (x2, y2)": str(test["p2"]),
        "Euclidean (L2)": dists["euclidean"],
        "Manhattan (L1)": dists["manhattan"],
        "Manhattan >= Euclidean": dists["manhattan"] >= dists["euclidean"]
    })

df_test_distances = pd.DataFrame(test_results)
print("=" * 85)
print("TASK 4: DUAL DISTANCE CALCULATION TEST RESULTS")
print("=" * 85)
print(df_test_distances.to_string(index=False))
```

## 5️⃣ Task 5: System Design: Linkage Choice for a 'Similar Movies' Engine

### 🎬 Scenario Context
You are designing the **"Similar Movies You May Enjoy"** recommendation carousels for a streaming application (like Netflix or Prime Video). Each movie is represented as a high-dimensional vector based on user rating patterns (collaborative filtering ratings matrix or latent embedding vectors from matrix factorization).

---

### ⚖️ Evaluation of Linkage Candidates for Movies

1. **Single Linkage — ❌ REJECT:**
   - **Fatal Flaw:** Suffers from extreme **chaining**. For example, a romantic comedy like *Crazy Rich Asians* might share a minor actor or soundtrack tag with an action movie like *Shang-Chi*, which connects to *Avengers*, chaining romance fans directly into hardcore superhero films without true user preference coherence.
   - Leads to giant, sprawling, heterogeneous movie buckets.

2. **Complete Linkage — ⚠️ SUB-OPTIMAL:**
   - **Flaw:** Requires every movie in a cluster to be close to every other movie (maximum distance). Highly polarizing cult movies (e.g., *Fight Club*, *Pulp Fiction*—which have intense 5-star and 1-star divides) will be completely excluded from their natural genres.

3. **Ward's Linkage — 🏆 PRIMARY RECOMMENDATION:**
   - **Mechanism:** Minimizes the total within-cluster variance (sum of squared deviations from the cluster centroid).
   - **Why It Wins:**
     - **Cohesive Taste Cohorts:** It builds compact, balanced, and evenly populated clusters. Movie fans receive tightly grouped sub-genres (e.g., *"90s Mind-Bending Sci-Fi Thrillers"* or *"Heartwarming Holiday Rom-Coms"*).
     - **Mathematically Clean:** Resembles K-Means' objective of variance minimization, but allows hierarchical browsing (a user can zoom into broad "Drama" $\rightarrow$ "Legal Drama" $\rightarrow$ "Courtroom True Crime").

4. **Average Linkage — 🥈 STRONG RUNNER-UP (Alternative for Sparse Rating Matrices):**
   - If computing pairwise similarity using **Cosine Distance** directly on raw, unnormalized user ratings (where Euclidean variance cannot be computed for Ward's method), **Average Linkage (UPGMA)** is the best alternative because it smooths out individual rating idiosyncrasies without chaining.

---

### 📝 Executive Summary:
> **Decision:** I would choose **Ward's Linkage** (or **Average Linkage** if using Cosine distance). Ward's method explicitly minimizes within-cluster variance, preventing disparate film categories from chaining together through bridge titles, and produces tight, evenly-sized movie micro-clusters with consistent viewer ratings.

## 🏁 Session 5 Summary & Takeaways

1. **Bottom-Up Agglomeration:** Starts from $N$ singleton clusters, consulting the pairwise distance matrix at every step to greedily fuse the closest clusters.
2. **Dendrogram Power:** Provides a complete visual hierarchy of all possible cluster granularities, eliminating the need to guess $k$ in advance.
3. **Linkage Criterion Dictates Cluster Geometry:**
   - `single` $\rightarrow$ Minimum distance (susceptible to chaining).
   - `complete` $\rightarrow$ Maximum distance (compact, diameter-constrained).
   - `average` $\rightarrow$ Mean distance (noise-resilient balance).
   - `ward` $\rightarrow$ Minimum variance (optimal for commercial recommendation engines).
4. **Metric Sensitivity:** Distance metrics ($L_1$ vs. $L_2$) govern the base geometry of space, directly influencing which points are prioritized during the initial merge steps.
