## 1️⃣ Task 1: Synthetic User Dataset Generation

### 📖 Conceptual Definition & Background
- **Unsupervised Feature Vectors:** In unsupervised learning, our dataset consists solely of feature vectors $\vec{x} \in \mathbb{R}^d$ without any target labels ($y$). The algorithm receives observations and seeks to discover inherent patterns, natural groupings, or distributions.
- **Feature Representation:**
  1. `daily_app_opens`: A discrete numerical count representing how frequently a user opens the mobile application within a 24-hour cycle.
  2. `avg_session_duration_mins`: A continuous numerical feature representing the average time in minutes a user spends active in the app per session.
- **Data Structure:** We represent the dataset as a **list of dictionaries**, the universal standard format for JSON-based web payloads, REST APIs, and document datastores (e.g., MongoDB), which is easily consumed by Pandas.

```python
# =====================================================================
# TASK 1: Generate a random dataset of 10 users
# =====================================================================
import random
import pandas as pd
from pprint import pprint

# Set random seed for reproducibility
random.seed(42)

# Generate dataset: 10 users with two features
user_dataset = []
for i in range(1, 11):
    user = {
        "user_id": f"User_{i:02d}",
        "daily_app_opens": random.randint(3, 40),               # Discrete count
        "avg_session_duration_mins": round(random.uniform(2.5, 45.0), 2)  # Continuous (mins)
    }
    user_dataset.append(user)

# Print as a list of dictionaries
print("=" * 65)
print("OUTPUT: Dataset as a List of Dictionaries (10 Users)")
print("=" * 65)
pprint(user_dataset, sort_dicts=False)

# Display as Pandas DataFrame for clear tabular visualization
df_users = pd.DataFrame(user_dataset)
print("\n" + "=" * 65)
print("TABULAR PREVIEW (Pandas DataFrame):")
print("=" * 65)
print(df_users.to_string(index=False))

# Quick statistical summary of the unsupervised features
print("\nDescriptive Statistics:")
display(df_users.describe()) if 'display' in globals() else print(df_users.describe())
```

### 🔍 Observation & Unsupervised Perspective:
- Each user $i$ is defined by a 2D coordinate: $(\text{daily\_app\_opens}_i, \text{avg\_session\_duration\_mins}_i)$.
- Notice there is **no target label** (e.g., whether the user is "Churn Risk" or "Power User").
- An unsupervised clustering algorithm (such as K-Means or DBSCAN) can segment these users into behavioral cohorts (e.g., *frequent quick checkers* vs. *rare deep engagers*).

## 2️⃣ Task 2: Supervised vs. Unsupervised Learning Comparison

### 📖 Conceptual Definition
- **Supervised Learning:** The model is presented with input features $X$ along with explicit ground-truth target labels $y$. The goal is to learn a mapping function $\hat{y} = f(X)$ by minimizing a loss function $\mathcal{L}(y, \hat{y})$.
- **Unsupervised Learning:** The model is presented only with input features $X$ without supervisor feedback or target labels. The goal is to uncover the underlying probability density $p(X)$, discover natural clusters $C$, or find a compressed low-dimensional manifold $Z$.

### 📊 Comprehensive Comparison Table

| Comparison Dimension | 🎯 Supervised Learning | 🔍 Unsupervised Learning |
| :--- | :--- | :--- |
| **1. Nature of Input Data** | **Labeled Data:** Requires pairs $(X, y)$ where every training sample has an associated ground truth target. | **Unlabeled Data:** Only feature matrix $X$ is provided; no target labels $y$ exist. |
| **2. Primary Objective** | **Prediction & Mapping:** Solve Classification (discrete labels) or Regression (continuous values) by mapping $y = f(X)$. | **Pattern & Structure Discovery:** Grouping (Clustering), Dimensionality Reduction, Anomaly Detection, or Association Rule Mining. |
| **3. Feedback & Guidance** | **Direct External Supervision:** Model errors are directly computed against true labels via loss functions (e.g., Cross-Entropy, MSE). | **Self-Directed Optimization:** Optimizes internal mathematical criteria (e.g., minimizing within-cluster inertia or maximizing variance). |
| **4. Evaluation Metrics** | **Ground-Truth Metrics:** Explicitly evaluated via Accuracy, Precision, Recall, F1-Score, ROC-AUC, Mean Squared Error (MSE). | **Internal Validation Metrics:** Evaluated via Silhouette Score, Davies-Bouldin Index, Calinski-Harabasz, or downstream business utility. |
| **5. Real-World App Example: Instagram** | **Feed & Reels Ranking:** Predicts the probability $P(\text{like}, \text{share}, \text{comment} \mid \text{user}, \text{post})$ using historical user engagement logs to rank posts. | **Aesthetic & Community Discovery:** Clusters hashtag embeddings, image visual vectors, and follow-graphs to detect emerging aesthetic subcultures without manual tags. |
| **6. Real-World App Example: Spotify** | **Track Skip Prediction:** Binary classification predicting whether a listener will skip a track within the first 30 seconds based on previous skip behavior. | **Discover Weekly Playlist Grouping:** Clusters songs into cohesive mood mixes using acoustic audio features (tempo, valence, danceability) and listening session co-occurrences. |
| **7. Real-World App Example: Zomato** | **Delivery ETA Prediction:** Regression model predicting exact delivery duration in minutes based on distance, traffic, and kitchen prep time history. | **Cloud Kitchen Hotspot Detection:** Spatial density clustering (DBSCAN) grouping delivery coordinates to pinpoint high-demand zones for new cloud kitchens. |

## 3️⃣ Task 3: Product Segmentation into Clusters (Price vs. Reviews)

### 📖 Problem Statement
We are given a product catalog of **8 products**, each characterized by two attributes:
1. **Price ($):** Monetary cost of the product.
2. **Number of Reviews:** Quantitative metric of sales volume, adoption, and social proof.

Our objective is to segment these products into **2 to 3 clusters** based on multidimensional similarity and provide clear analytical reasoning.

### 🛍️ The 8 Products Catalog

| ID | Product Name | Price ($) | Number of Reviews |
| :---: | :--- | :---: | :---: |
| P1 | Smartphone Screen Protector (3-Pack) | $10 | 12,500 |
| P2 | USB-C Braided Fast Charging Cable | $15 | 9,800 |
| P3 | Wireless Optical Office Mouse | $25 | 5,200 |
| P4 | RGB Mechanical Gaming Keyboard | $110 | 4,100 |
| P5 | Active Noise-Cancelling Headphones | $280 | 2,400 |
| P6 | 34-Inch Curved Ultra-Wide Gaming Monitor | $650 | 620 |
| P7 | Professional Studio Audio Reference Monitors | $1,150 | 95 |
| P8 | Full-Frame Mirrorless Cinema Camera | $2,400 | 45 |

---

### 🧠 Segmentation Methodology & Cluster Reasoning

We segment the products into **3 distinct clusters** along two business axes: **Affordability** vs. **Market Popularity / Social Proof**:

#### 🟢 Cluster 1: "Mass-Market Budget Essentials" (Low Price, High Popularity)
- **Members:** Smartphone Screen Protector ($10 / 12,500 rev), USB-C Fast Cable ($15 / 9,800 rev), Wireless Optical Mouse ($25 / 5,200 rev).
- **Price Range:** $10 – $25 (Ultra-affordable).
- **Review Range:** 5,200 – 12,500 (Massive volume of reviews).
- **Reasoning:** These are low-consideration, impulse everyday consumer commodities. Customers buy them frequently with minimal hesitation, generating thousands of reviews. They occupy the high-adoption, low-cost quadrant.

#### 🟡 Cluster 2: "Mid-Tier Consumer Tech & Peripherals" (Moderate Price, Moderate Popularity)
- **Members:** Mechanical Gaming Keyboard ($110 / 4,100 rev), Noise-Cancelling Headphones ($280 / 2,400 rev), Curved Gaming Monitor ($650 / 620 rev).
- **Price Range:** $110 – $650 (Considered consumer investments).
- **Review Range:** 620 – 4,100 (Solid, healthy social proof).
- **Reasoning:** These are discretionary lifestyle and productivity items. Buyers research before purchasing, meaning lower total sales volume than cables, but high engagement among enthusiasts. They represent the balanced middle segment.

#### 🔴 Cluster 3: "High-End Professional / Enterprise Equipment" (High Price, Niche / Low Popularity)
- **Members:** Studio Reference Monitors ($1,150 / 95 rev), Cinema Camera ($2,400 / 45 rev).
- **Price Range:** $1,150 – $2,400 (Commercial/Studio grade).
- **Review Range:** 45 – 95 (Exclusively professional/niche audience).
- **Reasoning:** These are high-barrier, capital expenditure tools for audio engineers, filmmakers, and studios. The target market is tiny compared to general consumers, leading to very low review counts despite elite price tags.

```python
# =====================================================================
# TASK 3: Product Segmentation & Clustering (Manual + K-Means + Plot)
# =====================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# 1. Define Product Dataset
products_data = [
    {"product_id": "P1", "product_name": "Screen Protector (3-Pack)", "price": 10, "reviews": 12500},
    {"product_id": "P2", "product_name": "USB-C Fast Cable", "price": 15, "reviews": 9800},
    {"product_id": "P3", "product_name": "Wireless Optical Mouse", "price": 25, "reviews": 5200},
    {"product_id": "P4", "product_name": "Mechanical Gaming Keyboard", "price": 110, "reviews": 4100},
    {"product_id": "P5", "product_name": "Noise-Cancelling Headphones", "price": 280, "reviews": 2400},
    {"product_id": "P6", "product_name": "Curved Gaming Monitor 34in", "price": 650, "reviews": 620},
    {"product_id": "P7", "product_name": "Studio Reference Monitors", "price": 1150, "reviews": 95},
    {"product_id": "P8", "product_name": "Cinema Mirrorless Camera", "price": 2400, "reviews": 45},
]
df_products = pd.DataFrame(products_data)

# 2. Manual Rule-Based Cluster Assignment
def manual_cluster(row):
    if row["price"] <= 50 and row["reviews"] >= 5000:
        return "Cluster 1: Budget Everyday Essentials"
    elif row["price"] <= 700 and row["reviews"] >= 500:
        return "Cluster 2: Mid-Tier Enthusiast Tech"
    else:
        return "Cluster 3: Professional / Niche Gear"

df_products["manual_cluster"] = df_products.apply(manual_cluster, axis=1)

# 3. Machine Learning Algorithmic Clustering (K-Means with StandardScaler)
# Note: Standardizing features is essential because price ($10-$2400) and reviews (45-12500) have vastly different scales!
features = ["price", "reviews"]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_products[features])

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_products["kmeans_cluster"] = kmeans.fit_predict(X_scaled)

print("=" * 80)
print("PRODUCT SEGMENTATION TABLE:")
print("=" * 80)
print(df_products[["product_id", "product_name", "price", "reviews", "manual_cluster", "kmeans_cluster"]].to_string(index=False))

# 4. Publication-Quality Visualization
fig, ax = plt.subplots(figsize=(10, 6))
colors = {"Cluster 1: Budget Everyday Essentials": "#2ecc71", 
          "Cluster 2: Mid-Tier Enthusiast Tech": "#f39c12", 
          "Cluster 3: Professional / Niche Gear": "#e74c3c"}

for cluster_name, group in df_products.groupby("manual_cluster"):
    ax.scatter(group["price"], group["reviews"], label=cluster_name, color=colors[cluster_name], s=140, edgecolors="black", zorder=3)

# Annotate product names on the scatter plot
for _, row in df_products.iterrows():
    ax.annotate(
        f"{row['product_id']}: {row['product_name']}",
        (row["price"], row["reviews"]),
        textcoords="offset points",
        xytext=(8, 6),
        fontsize=9,
        fontweight="semibold"
    )

ax.set_title("Product Segmentation: Price vs. Number of Reviews", fontsize=14, fontweight="bold", pad=12)
ax.set_xlabel("Price ($)", fontsize=11, fontweight="bold")
ax.set_ylabel("Number of Reviews (Popularity / Social Proof)", fontsize=11, fontweight="bold")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend(title="Segments", frameon=True, facecolor="#f8f9fa", loc="upper right")
plt.tight_layout()
plt.show()
```

## 4️⃣ Task 4: Unsupervised Learning Scenarios in Zomato, Flipkart, and Spotify

Unsupervised learning is crucial in consumer internet platforms because user behaviors are multifaceted, unlabelled, and dynamic. Below are 3 in-depth scenarios:

---

### 🍕 Scenario 1: Zomato — Delivery Hotspot & Cloud Kitchen Location Clustering
- **Business Challenge:** How to determine where Zomato should establish new "Zomato Kitchens" (cloud kitchen infrastructure) and where to stage delivery riders ahead of peak dinner rushes.
- **Data Needed:**
  1. Geospatial order coordinates (customer drop latitude and longitude).
  2. Temporal timestamps (order time, day of week, rain/weather flags).
  3. Cuisine demands (e.g., North Indian, Biryani, Desserts, Healthy Salads).
  4. Delivery transit delays and kitchen prep turnaround times.
- **Unsupervised Technique:** **Density-Based Spatial Clustering of Applications with Noise (DBSCAN / HDBSCAN)**.
- **Hidden Patterns & Insights Found:**
  - Discovers high-density clusters of unsatisfied demand (e.g., a massive cluster of late-night food orders in a university hostel district with average delivery times > 50 minutes due to lack of local restaurants).
  - Pinpoints the exact mathematical centroid for a cloud kitchen facility to minimize average delivery radius down to < 2.5 km, slashing delivery time and fuel costs.

---

### 🛒 Scenario 2: Flipkart — Cross-Category "Market Basket" & Buyer Journey Persona Discovery
- **Business Challenge:** E-commerce catalogs are organized into rigid categories (e.g., Electronics vs. Grocery), but human life events span across multiple categories simultaneously.
- **Data Needed:**
  1. Co-purchased item session logs (items checked out in the same transaction or within a 14-day window).
  2. Clickstream browsing history (dwell time, zoom-in on images, search query chains).
  3. Price sensitivity indicators (use of EMI options, discount coupon hunting, brand tier preferences).
- **Unsupervised Technique:** **Association Rule Mining (FP-Growth / Apriori)** combined with **Agglomerative Hierarchical Clustering** on customer embedding vectors.
- **Hidden Patterns & Insights Found:**
  - Discovers latent life-stage clusters (e.g., the *"New Pet Parent"* cluster purchasing dog food, stain removers, baby safety gates, and chew toys across 3 separate store departments).
  - Discovers *"WFH Setup Renovator"* buying ergonomics chairs, monitor riser stands, and blue-light glasses.
  - Drives automated cross-category recommendation widgets ("Customers also bundled these") without manual curation by catalog managers.

---

### 🎵 Scenario 3: Spotify — Acoustic Micro-Mood Clustering ("Discover Weekly" Engine)
- **Business Challenge:** Human genre tags ("Pop", "Indie", "Rock") are notoriously broad and subjective. A listener looking for focus music does not care about genre; they care about sound texture, tempo, and mood.
- **Data Needed:**
  1. Low-level DSP acoustic features: Tempo (BPM), Energy (0 to 1), Valence (musical positiveness, 0 to 1), Danceability, Acousticness, Instrumentalness, Speechiness.
  2. Deep audio spectrogram embeddings extracted from raw audio waveforms.
  3. Session context: Time of listening (morning workout vs. 2 AM study session) and skip events within the first 15 seconds.
- **Unsupervised Technique:** **Gaussian Mixture Models (GMM) / K-Means** coupled with **t-SNE / UMAP** dimensionality reduction on 512-dimensional acoustic embeddings.
- **Hidden Patterns & Insights Found:**
  - Automatically clusters tracks into micro-mood acoustic spaces (e.g., *"Late-Night Melancholy Lo-Fi"*, *"High-Adrenaline Cardio Synthwave"*, *"Subdued Acoustic Coffeehouse"*).
  - Allows Spotify to recommend lesser-known independent artists whose audio profile is virtually identical to popular tracks in a user's cluster, even if the artist has zero genre tags or mainstream fame.

### 📐 Dataset Structure & Metadata Description
- **Total Samples ($N$):** 200 customer entries ($200 \times 5$ matrix).
- **Total Number of Features:** **5 features**.
- **Type of Data:** Tabular structured data containing mixed data types (Categorical, Discrete Numerical, and Continuous Numerical).

#### Feature-by-Feature Breakdown:
1. `CustomerID`: **Nominal / Integer Identifier** (e.g., 1, 2, ..., 200). Serves strictly as a primary key; excluded during clustering as it carries no behavioral information.
2. `Gender`: **Nominal Categorical** (`Male`, `Female`). Demographic binary variable.
3. `Age`: **Discrete Numerical Ratio Scale** (Range: 18 – 70 years). Customer age in years.
4. `Annual Income (k$)`: **Continuous Numerical** (Range: $15,000 – $137,000). Represents the customer's yearly earning power.
5. `Spending Score (1-100)`: **Continuous / Ordinal Numerical** (Range: 1 – 99). A proprietary behavioral index assigned by the shopping mall management based on purchase frequency, monetary volume, and loyalty card swipes.

---

### ❓ Is It Labeled or Unlabeled?
- **STRICTLY UNLABELED:**
- **Why?** There is **no predefined target column ($y$)**. No column tells us whether a customer belongs to "High Spender", "At-Risk", "VIP", or "Casual Shopper".
- The data science objective is 100% **unsupervised exploratory clustering**: to feed the feature space (typically `Age`, `Annual Income`, and `Spending Score`) into clustering algorithms (like K-Means, Agglomerative Hierarchical, or DBSCAN) to discover natural customer personas.

#### Typical 5 Natural Clusters Discovered in this Dataset:
1. **High Income, High Spending Score:** *"Target / VIP Shoppers"* (Mall's most lucrative cohort; prioritize luxury offerings).
2. **High Income, Low Spending Score:** *"Careful / Conservative Shoppers"* (High purchasing power but low engagement; target with premium marketing).
3. **Low Income, High Spending Score:** *"Careless / Trend Seekers"* (Younger demographic spending beyond income on trending products).
4. **Low Income, Low Spending Score:** *"Sensible / Frugal Shoppers"* (Price-sensitive, budget shoppers).
5. **Medium Income, Medium Spending Score:** *"Standard / Average Shoppers"* (Balanced shoppers, representing the mall's steady baseline).
